# Exercicio par/impar

```python
def par_ou_impar(n):
    if n % 2 == 0 :
        return 'Par'
    else :
        return 'Impar'


print ('O número 4 é', par_ou_impar(4))
print('O núemro 5 é', par_ou_impar(5))

```

## Exercicio Fatorial

```python
def fatorial(n):
    if n < 0:
        return 'Número invalido'
    elif n == 0:
        return 1
    else :
        resultado = 1
        for i in range (1, n + 1):
            resultado *= i
        return resultado

print('O fatorial de -1', fatorial(-1))
    
```

# Exercicio Lista

```python
def maior_elemento(lista):
    return max(lista)

numeros = [1, 5, 100 ,500, 25]
print('O maior elemento da lista é', maior_elemento(numeros))
```

## Exercico Função Aritimetica

```python
def media_a(lista):
    return sum(lista) / len(lista)


valores = [100, 250, 600, 750, 800]
print('O valor médio é', media_a(valores))
```

```python
def eh_palindromo(palavra):
    return palavra == palavra[::-1]
# Exemplos de chamada
print("ana é palíndromo?", eh_palindromo("ana"))
print("python é palíndromo?", eh_palindromo("python"))
```

```python
def contador_vogais(texto):
    vogais = 'aeiouAEIOU'
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1
    return contador
#
print('Números de vogais em "Python": ', contador_vogais("Python"))
```

```python
def ordenar_por_segundo(lista_tuplas):
    return sorted(lista_tuplas, key=lambda x: x[1])
# Exemplo de chamada
lista_exemplo = [("a", 3), ("b", 1), ("c", 2)]
print("Lista ordenada:", ordenar_por_segundo(lista_exemplo))
```
