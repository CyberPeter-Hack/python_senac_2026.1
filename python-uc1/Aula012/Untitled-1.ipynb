```python
class Carro:
    # Método construtor (__init__) para inicializar os atributos
    def __init__(self, marca, modelo, portas=4, ano='2026', cor='branca'):
        self.cor = cor          # Atributo
        self.tp_tracao = 2         # Atributo
        self.marca = marca      # Atributo
        self.modelo = modelo    # Atributo
        self.ano = ano          # Atributo
        self._ligado = False     # Atributo com valor padrão
        self.dir_hidraulica = True
        self.qt_portas = portas
        self.pontecia = 1.0
        self._cambio_automatico = False
        self._4p4 = False

    # Método para ligar o carro
    def ligar(self):
        if not self._ligado:
            self._ligado = True
            print(f"{self.marca} {self.modelo} está ligado.")
        else:
            print(f"{self.marca} {self.modelo} já está ligado.")

    # Método para desligar o carro
    def desligar(self):
        if self._ligado:
            self._ligado = False
            print(f"{self.marca} {self.modelo} está desligado.")
        else:
            print(f"{self.marca} {self.modelo} já está desligado.")

        # Método para desligar o 4x4
    def ligar(self):
        if not self._cambio_automatico:
            self._cambio_automatico = True
            print(f"{self.marca} {self.modelo} está com 4x4 ligado.")
        else:
            print(f"{self.marca} {self.modelo} já está com 4x4 ligado.")

    # Método de ligar o 4x4
    def desligar(self):
        if self._cambio_automatico:
            self._cambio_automatico = False
            print(f"{self.marca} {self.modelo} está com 4x4 desligado.")
        else:
            print(f"{self.marca} {self.modelo} já está com 4x4 desligado.")


    # Método para exibir informações do carro
    def exibir_info(self):
        status = "ligado" if self._ligado else "desligado"
        print(f"{self.marca} {self.modelo} ({self.ano}) está {status}.")

    def muda_cor (self):
        self.cor=cor
        return True

    def mudar_tracao (self):
        if self.tipo.tracao == 2:
            self.tp_tracao = 4
        else : 
            self.tp_tracao = 2
        return True

    def mudar_potencia (self, pot='1.0'):
        try:
            vl_pot=float(pot)
            self.potencia=vl_pot
        except:    
            return True



    
# Criando um objeto da classe Carro
meu_carro = Carro("Toyota", "Corolla", 4, 2020)

# Usando métodos do objeto
meu_carro.exibir_info()  # Exibe as informações do carro
meu_carro.ligar()  # Liga o carro
meu_carro.exibir_info()  # Exibe as informações do carro novamente
meu_carro.desligar()  # Desliga o carro
meu_carro.exibir_info()  # Exibe as informações do carro novamente



```
