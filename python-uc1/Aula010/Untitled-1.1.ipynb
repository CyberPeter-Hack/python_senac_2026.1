```python
import random

def criar_matriz(qt_linhas=4, qt_colunas=4):
    matriz=[]
    for _linha in range (qt_linhas):
        linha=[]
        for _coluna in range (qt_colunas):
            linha.append(random.randint(1,100))
        matriz.append(linha)
    return (matriz)

matriz=criar_matriz(5,5)

print(matriz)
```

```python
def imprime_matriz(matriz=[]):
    for linha in matriz:
        for elemento in linha:
            print(f'{elemento:4d}', end='')
        print()
        
```

```python
minha_matriz=criar_matriz(6,7)
```

```python
imprime_matriz(minha_matriz)
```
