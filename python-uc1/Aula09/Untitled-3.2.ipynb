```python
produtos = (
    ('Arroz', 5.99, 3),
    ('Carne', 25.99, 4),
    ('Leite', 10.99, 6),
    ('Feijão', 12.99, 3),
    ('Óleo', 9.99, 5)
)
valor_produto = 0
valor_total = 0
print('Lista de Produtos')
for nome, preco, qt in produtos:
    valor_produto = preco * qt
    valor_total += valor_produto 
    print(f'{nome:.<20} R$ {preco:.2f}, Qt {qt:02d}, total R${valor_produto}')
print(f'\n\t O valor total da compra é R${valor_total:7.2f}')
```

```python
numeros = (3, 12, 52, 96, 7, 5, 10, 99)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


primos = tuple(num for num in numeros if eh_primo(num))
print(f"Números primos: {primos}")

```

```python
import random
idades=[]
for _ in range (100):
    idades.append(random.randint(1,92))


maiores=0
for idade in idades:
    if idade>=18:
        maiores+=1

print(f'Quantidade de maiores de idade: {maiores}')
```
