```python
def potencia (numero, expoente):
    resultado = numero ** expoente
    return resultado
```

```python
numero=2
expoente=8

print(potencia(numero, expoente))
```

# Argumento Arbitrário

```python
def saudacao(display='sem nome'):
    print (f'Ola, {display} !!!')
```

```python
saudacao('Papagaio')
```

```python
saudacao()
```

## Passagem de argumentos arbitrarios

```python
def saudacoes (*nomes):
  #  print(nomes)
    for nome in nomes:
        saudacao(nome)
```

```python
saudacoes ('Luis', 'Rodrigo', 'Papagaio', 'Guilherme')
```
