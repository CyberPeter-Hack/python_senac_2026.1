```python
produtos = (
    ('Arroz', 5.99),
    ('Carne', 25.99),
    ('Leite', 10.99),
    ('Feijão', 12.99),
    ('Óleo', 9.99)
)

print('Lista de Produtos')
for nome, preco in produtos:
    print(f'{nome:.<20} R$ {preco:.2f}')
```

```python
notas = (7.5, 8.5, 9, 5.5)
media = sum(notas) / len(notas)
print(f'A média das notas é {media:.2f}')
```

```python
numeros = (3, 12, 52, 96, 7, 5, 10, 99)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


primos = tuple(num for num in numeros if eh_primo(num))
print(f"Números primos: {primos}")
```

```python
idades = (15, 25, 18, 19, 12)
maior_idade = sum(1 for idade in idades if idade >= 18)
print(f'A quantidade de maiores de 18 anos é {maior_idade}')
```
